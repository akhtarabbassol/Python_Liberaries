```python
a = np.array([10,10,20,20,30,30])

remove_dup = []

for i in range(len(a)):
    for j in range(i+1,len(a)):
        if a[i] == a[j]:
            remove_dup.append(a[j])

print(remove_dup)
```

```python
import numpy as np
#a = np.array([7,6,5,6,1,2,3])
a = np.random.rand(10)
print('Before Sorting,', a)
for i in range(len(a)):
    for j in range(0, len(a)-i-1):
        if a[j] > a[j+1]:
            a[j] , a[j+1] = a[j+1] , a[j]


print('After Sorting', a)
```
