## 1. Which month saw the highest joining of the employees

   function name: max_joinings_month()
   
   inputs: DataFrame
   
   outputs: str

```python
df[["Month Name of Joining","Month of Joining"]].head()
```

```python
def max_joining_month(df):
    highest_joining = df['Month Name of Joining'].value_counts()
    month =  highest_joining.idxmax()
    joining = highest_joining.max()
    return f"{month}, has the highest number of joining which is  {joining}"
max_joining_month(df)
```

## 2. What is the mean salary is the state KS
   
   function name: mean_ks_salary()
   
   inputs: DataFrame
   
   outputs: float

```python
def mean_ks_salary(df):
    mean_salary = df[df['State']=='KS']['Salary'].mean()
    return mean_salary
mean_ks_salary(df)
```

## 3. Which Region has highest mean salary
   
   function name: max_mean_salary_region()
   
   inputs: DataFrame
   
   outputs: str

```python
def max_mean_salary_region(df):
    max_region = df.groupby('Region')['Salary'].mean()
    region = max_region.idxmax()
    salary = max_region.max()
    return f"{region} have the highest mean salary {salary}"
max_mean_salary_region(df)
```

## 4. Which region has lowest mean salary
   
   function name: min_mean_salary_region()
   
   inputs: DataFrame
   
   outputs: str

```python
def min_mean_salary_region(df):
    min_region = df.groupby('Region')['Salary'].mean()
    region = min_region.idxmin()
    salary = min_region.min()
    return f"{region} have the lowest mean salary {salary}"
min_mean_salary_region(df)
```

## 5. Which City has highest mean weight
   
   function name: max_mean_weight_city()
   
   inputs: DataFrame
   
   outputs: str

```python
def max_mean_weight_city(df):
    max_weight = df.groupby('City')['Weight in Kgs.'].agg('mean')
    return max_weight.idxmax(), max_weight.max()
max_mean_weight_city(df)
```

## 6. What is the difference between mean salary of Doctors and Professors
   
   function name: dr_prof_mean_salary_diff()
   
   inputs: DataFrame
   
   outputs: float

```python
df[['Name Prefix']].to_string()
```

```python
def dr_prof_mean_salary_diff(df):
    dr_mean_salary = df[df['Name Prefix']=='Drs.']['Salary'].mean()
    prof_mean_salary = df[df['Name Prefix']=='Prof.']['Salary'].mean()
    diff = dr_mean_salary-prof_mean_salary
    return diff
dr_prof_mean_salary_diff(df)
```

## 7. How many female use yahoo
   
   function name: females_with_yahoo()
   
   inputs: DataFrame
   
   outputs: int

```python
def females_with_yahoo(df):
    yahoo = df[(df['Gender']=='F') &(df['E Mail'].str.split('@').str[-1]=='yahoo.com')]
    return len(yahoo)
females_with_yahoo(df)
```

```python
def females_with_yahoo(df):
    yahoo = df[(df['Gender']=='F') &(df['E Mail'].str.endswith('@yahoo.com', na = False))]
    return len(yahoo)
females_with_yahoo(df)
```

## 8. How many Doctors use gmail
   
   function name: doctors_with_gmail()
   
   inputs: DataFrame
   
   outputs: int

```python
def doctors_with_gmail(df):
    gmail = df[(df['Name Prefix']=='Drs.') & (df['E Mail'].str.endswith('@gmail.com', na = False))]
    return len(gmail)
doctors_with_gmail(df)
```

## 9. What is the password of the youngest male
   
   function name: youngest_male_password()
   
   inputs: DataFrame
   
   outputs: str

```python
def youngest_male_password(df):
    male = df[df['Gender'] == 'M']
    password = male.loc[male['Age in Yrs.'].idxmin(), ['Password']]
    return password
youngest_male_password(df)
```
